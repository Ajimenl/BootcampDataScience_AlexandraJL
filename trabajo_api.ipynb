```python
import requests
base_url = "https://api-colombia.com/"
endpoint = "/api/v1/Airport"

respuesta_api = requests.get(base_url + endpoint)
```

```python
import pandas as pd 
print(respuesta_api.status_code)
respuesta_json = respuesta_api.json()

df_api = pd.DataFrame.from_dict(respuesta_json)
```

```python
df_api.to_csv("aeropuertos.csv",index= False)
```
